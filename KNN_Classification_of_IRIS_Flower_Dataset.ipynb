```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, pred)
print(cm)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, pred)

print(report)
```

```python
k_range=range(1,40)
scores=[]
for k in k_range:
    knn= KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train,y_train)
    accuracyScore=knn.score(x_test,y_test)
    scores.append(accuracyScore)
print(scores)
plt.figure()
plt.xlabel("K Value")
plt.ylabel("Accuracy Score")
plt.scatter(k_range,scores)
plt.show()
```

## **Decision Tree Classification**


```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(x_train, y_train)
y_pred_dt = dt_model.predict(x_test)

print("Decision Tree Accuracy:", dt_model.score(x_test, y_test))
```

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)
print(cm_dt)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(cm_dt, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Decision Tree Confusion Matrix')
plt.show()
```

```python
report_dt = classification_report(y_test, y_pred_dt)
print(report_dt)
```

```python
k_range_dt = range(1, 40)
scores_dt = []

for k in k_range_dt:
    dt_model = DecisionTreeClassifier(max_depth=k)
    dt_model.fit(x_train, y_train)
    accuracy_score_dt = dt_model.score(x_test, y_test)
    scores_dt.append(accuracy_score_dt)
```

```python
plt.figure()
plt.xlabel("Max Depth")
plt.ylabel("Accuracy Score")
plt.scatter(k_range_dt, scores_dt)
plt.show()
```

## **Random Forest**

```python
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)
y_pred_rf = rf_model.predict(x_test)

print("Random Forest Accuracy:", rf_model.score(x_test, y_test))

cm_rf = confusion_matrix(y_test, y_pred_rf)
print(cm_rf)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Random Forest Confusion Matrix')
plt.show()
```

```python
report_rf = classification_report(y_test, y_pred_rf)
print(report_rf)
```

```python
k_range_rf = range(1, 40)
scores_rf = []
for k in k_range_rf:
    rf_model = RandomForestClassifier(n_estimators=k)
    rf_model.fit(x_train, y_train)
    accuracy_score_rf = rf_model.score(x_test, y_test)
    scores_rf.append(accuracy_score_rf)
plt.figure()
plt.xlabel("Number of Estimators")
plt.ylabel("Accuracy Score")
plt.scatter(k_range_rf, scores_rf)
plt.show()
```
